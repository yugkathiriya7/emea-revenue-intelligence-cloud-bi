```python

# Preserve original raw data
df_clean = df_raw.copy()

# 1. Remove exact duplicate rows
before = len(df_clean)
df_clean = df_clean.drop_duplicates().copy()

print("Duplicates removed:", before - len(df_clean))

# 2. Standardise currency codes
df_clean["currency"] = (
    df_clean["currency"]
    .astype("string")
    .str.strip()
    .str.upper()
    .replace({"EURO": "EUR"})
)

# 3. Check remaining currencies
print("\nCLEANED CURRENCY DISTRIBUTION")
display(df_clean["currency"].value_counts(dropna=False))

# 4. Verify invoice uniqueness
print("\nRemaining duplicate invoice IDs:",
      df_clean["invoice_id"].duplicated().sum())

# 5. Verify invalid currencies
print("Invalid currencies:",
      (~df_clean["currency"].isin(["EUR", "GBP"])).sum())

print("\nCleaned records:", len(df_clean))
```

```python

master = client.query("""
SELECT *
FROM `emea-strategic-analytics.emea_analytics.account_master`
""").to_dataframe()

print("ACCOUNT MASTER COLUMN NAMES:")
print(master.columns.tolist())

print("\nFIRST 10 MASTER RECORDS:")
display(master.head(10))
```

```python

# STEP 12: Resolve ERP customer identities
# using the governed account master

# Ensure one approved master record per account ID
assert master["account_id"].is_unique

# Bring master details into the cleaned billing dataset
df_enriched = df_clean.merge(
    master[[
        "account_id",
        "canonical_account_name",
        "parent_account_id",
        "home_cluster"
    ]],
    left_on="erp_customer_code",
    right_on="account_id",
    how="left",
    validate="many_to_one",
    indicator=True
)

# Identify records that could not be mapped
unmatched = df_enriched[
    df_enriched["_merge"] != "both"
].copy()

# Use the approved master name for matched records
df_enriched["resolved_customer_name"] = (
    df_enriched["canonical_account_name"]
)

# Flag missing or inconsistent source names
source_name = (
    df_enriched["erp_customer_name"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.casefold()
)

master_name = (
    df_enriched["canonical_account_name"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.casefold()
)

df_enriched["customer_name_mismatch"] = (
    source_name != master_name
)

print("Total invoice records:", len(df_enriched))
print("Mapped records:",
      (df_enriched["_merge"] == "both").sum())
print("Unmatched records:", len(unmatched))
print("Source name differences:",
      df_enriched["customer_name_mismatch"].sum())

display(
    df_enriched[[
        "invoice_id",
        "erp_customer_code",
        "erp_customer_name",
        "resolved_customer_name",
        "customer_name_mismatch"
    ]].head(10)
)
```

```python

# Load exchange rates from BigQuery
fx_data = client.query("""
SELECT *
FROM `emea-strategic-analytics.emea_analytics.fx_rates`
""").to_dataframe()

print("FX TABLE COLUMNS")
print(fx_data.columns.tolist())

print("\nFX SAMPLE")
display(fx_data.head(10))
```

```python

# STEP 16: Monthly currency conversion

import pandas as pd
import numpy as np

# Create a working copy
df_fx = df_enriched.copy()

# Convert invoice dates into monthly keys
df_fx["invoice_date"] = pd.to_datetime(
    df_fx["invoice_date"]
)

df_fx["fx_month"] = (
    df_fx["invoice_date"]
    .dt.strftime("%Y-%m")
)

# Prepare monthly FX rates
rates = fx_data.copy()

rates["fx_month"] = (
    rates["fx_month"]
    .astype(str)
    .str[:7]
)

rates["from_currency"] = (
    rates["from_currency"]
    .astype(str)
    .str.strip()
    .str.upper()
)

rates = rates[
    rates["to_currency"] == "EUR"
].copy()

# Prevent duplicated FX matches
assert not rates.duplicated(
    ["fx_month", "from_currency"]
).any(), "Duplicate monthly FX rates detected"

# Join invoices to monthly exchange rates
df_fx = df_fx.merge(
    rates[[
        "fx_month",
        "from_currency",
        "eur_per_unit"
    ]],
    left_on=["fx_month", "currency"],
    right_on=["fx_month", "from_currency"],
    how="left",
    validate="many_to_one"
)

# Calculate local and EUR revenue
df_fx["revenue_local"] = (
    pd.to_numeric(df_fx["quantity"]) *
    pd.to_numeric(df_fx["unit_price_local"])
)

df_fx["revenue_eur"] = (
    df_fx["revenue_local"] *
    pd.to_numeric(df_fx["eur_per_unit"])
).round(2)

# Flag records that cannot be reliably reported
df_fx["rejection_reason"] = ""

df_fx.loc[
    df_fx["account_id"].isna(),
    "rejection_reason"
] += "Unmatched account; "

df_fx.loc[
    df_fx["eur_per_unit"].isna(),
    "rejection_reason"
] += "Missing FX rate; "

df_fx.loc[
    df_fx["revenue_local"].isna() |
    (df_fx["revenue_local"] <= 0),
    "rejection_reason"
] += "Invalid revenue; "

# Split accepted and rejected records
fact_revenue = df_fx[
    df_fx["rejection_reason"] == ""
].copy()

rejected_records = df_fx[
    df_fx["rejection_reason"] != ""
].copy()

print("Original raw records:", len(df_raw))
print("After deduplication:", len(df_clean))
print("Accepted invoices:", len(fact_revenue))
print("Quarantined invoices:", len(rejected_records))

print("\nTotal accepted revenue EUR:",
      round(fact_revenue["revenue_eur"].sum(), 2))

print("\nQuarantined records:")
display(rejected_records[[
    "invoice_id",
    "erp_customer_code",
    "currency",
    "rejection_reason"
]])

print("\nSample validated revenue:")
display(fact_revenue[[
    "invoice_id",
    "resolved_customer_name",
    "currency",
    "revenue_local",
    "eur_per_unit",
    "revenue_eur"
]].head())
```

```python

# Reconciliation checks

assert (
    len(fact_revenue) + len(rejected_records)
    == len(df_clean)
), "Record reconciliation failed!"

assert fact_revenue["invoice_id"].is_unique, (
    "Duplicate invoice IDs in fact table!"
)

assert fact_revenue["revenue_eur"].notna().all(), (
    "Missing converted revenue!"
)

print("RECONCILIATION PASSED")
print("Accepted:", len(fact_revenue))
print("Rejected:", len(rejected_records))
print("Total:", len(df_clean))
```

```python

# STEP 18: Load transformed data into BigQuery

from google.cloud import bigquery

project_id = "emea-strategic-analytics"
dataset_id = "emea_analytics"

# Select clean reporting columns
revenue_to_upload = fact_revenue[[
    "invoice_id",
    "invoice_date",
    "account_id",
    "resolved_customer_name",
    "parent_account_id",
    "home_cluster",
    "product_code",
    "product_name",
    "contract_id",
    "currency",
    "quantity",
    "unit_price_local",
    "fx_month",
    "eur_per_unit",
    "revenue_local",
    "revenue_eur"
]].copy()

# Standardise dates and numeric types
revenue_to_upload["invoice_date"] = pd.to_datetime(
    revenue_to_upload["invoice_date"]
).dt.date

for col in [
    "quantity", "unit_price_local",
    "eur_per_unit", "revenue_local", "revenue_eur"
]:
    revenue_to_upload[col] = pd.to_numeric(
        revenue_to_upload[col]
    ).astype("float64")

# Keep a focused rejected-record audit table
rejected_to_upload = rejected_records[[
    "invoice_id",
    "erp_customer_code",
    "erp_customer_name",
    "currency",
    "rejection_reason"
]].copy()

# BigQuery upload configuration
job_config = bigquery.LoadJobConfig(
    write_disposition="WRITE_TRUNCATE"
)

# Upload accepted invoices
revenue_table = (
    f"{project_id}.{dataset_id}.fact_revenue"
)

job = client.load_table_from_dataframe(
    revenue_to_upload,
    revenue_table,
    job_config=job_config
)
job.result()

print("Uploaded:", revenue_table)
print("Rows:", len(revenue_to_upload))

# Upload rejected invoices
rejected_table = (
    f"{project_id}.{dataset_id}.rejected_records"
)

job = client.load_table_from_dataframe(
    rejected_to_upload,
    rejected_table,
    job_config=job_config
)
job.result()

print("Uploaded:", rejected_table)
print("Rows:", len(rejected_to_upload))
```
